```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# c = x + iy
x = np.linspace(-2, 2, 100)
y = np.linspace(-2, 2, 100)
X,Y = np.meshgrid(x,y)
c = X + 1j * Y
```

```python
def convergence(c, max_iter):
    z = np.zeros(c.shape)
    diverge_at = np.zeros(c.shape, dtype)
    
    for n in range(max_iter):
        conv = np.abs(z) < 2
        z[conv] = (z[conv])**2 + c[conv]
        diverge_at[conv] += 1
    
    return diverge_at
```

```python
plt.contour(X, Y, convergence(c, 100), levels=500)
plt.colorbar()
plt.show
```

```python
z = c
z = z**2 + c
z = z**2 + c
plt.contour(X, Y, abs(z), levels=500)
plt.colorbar()
plt.show
```
