# Part 1

```python
import math
import numpy
```

```python
def f(x: float) -> float:
    """Calculate the value of f(x) = x^3 - x^2 - x for some value x
    """
    return x**3 - x**2 - x
```

```python
def df(x: float) -> float:
    """Calculate the derivative of the function f(x) = x^3 - x^2 - x for some value x
    """
    return 3 * x**2 - 2 * x - 1
```

# Part 2

```python
def newton(f, df, x0: float, epsilon: float = 1e-6, max_iter: int = 30) -> float | None:
    """Finds the root of a function `f` 
    
    Parameters
    ----------
    f
        Function for which we are trying to find the root of
    df
        Function that gives the derivative of `f`
    x0: float
        Initial value from which we are trying to find the root
    epsilon: float, optional
        Error bound to determine whether or not the root has been found
    max_iter: int, optional
        The number of iterations to run before deciding to exit
        
    Returns
    -------
    float
        The value of x where f(x) = 0
    None
        If `max_iter` is exceeded and the program fails to find the root
    """
    xn = x0
    n = 0
    while n < max_iter and abs(f(xn)) >= epsilon:
        xn = xn - f(xn)/df(xn)
        n += 1
    
    if f(xn) < epsilon:
        print("Found root in", n, "interations")
        return xn
    else:
        print("Iteration failed")
        return None
```

# Part 3

```python
print(newton(f, df, 0))
print(newton(f, df, 2))
print(newton(f, df, -2))
# print(newton(f, df, 1))
```

```python
print (newton(f, df, 2, 1e-8))
```

For x0 = 2, it takes one more iteration to converge when epsilon = 1e-8
x0 = 1 yields a `ZeroDivisionError` 

```python
print (newton(f, df, 2, 1e-10))
print (newton(f, df, 2, 1e-20))
print (newton(f, df, 2, 1e-50))
print (newton(f, df, 2, 1e-324))
```
